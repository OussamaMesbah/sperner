# Nash equilibria

Every finite game has an equilibrium in mixed strategies (Nash 1950). Nash's proof writes one down
as a fixed point of a continuous map, so Brouwer's theorem — and the walk — find it.

```python
from sperner.nash import equilibrium, nash_map, symmetric_equilibrium

# Two cats at a bowl: hiss or wait. Food worth 4, a fight costs 6.
cats = [[(4 - 6) / 2, 4], [0, 4 / 2]]
result = symmetric_equilibrium(cats)
print("hiss with probability", round(result.strategy[0], 6), "- food / fight =", 4 / 6)
print("regret:", result.regret)
```

## Nash's map

The fixed points of Nash's map are the equilibria. At the equilibrium it leaves the strategy
alone; elsewhere it nudges towards what does better.

```python
f = nash_map(cats)
for p in (0.2, 2 / 3, 0.9):
    print(f"hiss {p:.3f} -> {f([p, 1 - p])[0]:.3f}")
```

## Two different players

```python
battle = equilibrium([[2, 0], [0, 1]], [[1, 0], [0, 2]])
print("battle of the sexes:", [round(v, 4) for v in battle.row + battle.column])
pennies = equilibrium([[1, -1], [-1, 1]], [[-1, 1], [1, -1]])
print("matching pennies:", [round(v, 4) for v in pennies.row + pennies.column])
```
