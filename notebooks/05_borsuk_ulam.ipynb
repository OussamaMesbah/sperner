# Tucker's lemma and the Borsuk–Ulam theorem

Every continuous map from the sphere to the plane sends two opposite points to the same place: two
antipodal places on Earth have the same temperature and pressure. The proof goes through Tucker's
lemma, the antipodal cousin of Sperner's lemma.

```python
import math

from sperner.tucker import antipodal_pair


def weather(p):
    x, y, z = p
    lat, lon = math.asin(z), math.atan2(y, x)
    temperature = 28 - 40 * (z - 0.2) ** 2 + 8 * math.sin(2 * lon + 1) * math.cos(lat)
    pressure = 1013 + 12 * math.sin(lon + 2) * math.cos(lat) + 7 * math.sin(4 * lat)
    return temperature, pressure


pair = antipodal_pair(weather, k=128)
opposite = tuple(-v for v in pair.point)
print("here:    ", [round(v, 3) for v in pair.point], [round(v, 2) for v in weather(pair.point)])
print("opposite:", [round(v, 3) for v in opposite], [round(v, 2) for v in weather(opposite)])
```

The two readings agree up to the grid: halving the grid spacing halves the guaranteed bound on
the difference, and in the limit they are equal. The actual difference shrinks roughly with the
spacing, though not always. Try `k=32` and `k=256`.

## Tucker's lemma

Label the corners of a symmetrically triangulated square with ±1, ±2 so that opposite boundary
points get opposite labels. Then some edge joins opposite labels.

```python
import random

from sperner.tucker import complementary_edge

k = 4
rng = random.Random(0)
labels = {}
for i in range(-k, k + 1):
    for j in range(-k, k + 1):
        if (i, j) not in labels:
            labels[(i, j)] = rng.choice([1, -1, 2, -2])
            if max(abs(i), abs(j)) == k:
                labels[(-i, -j)] = -labels[(i, j)]
u, v = complementary_edge(k, labels.__getitem__)
print("complementary edge:", u, v, "labels:", labels[u], labels[v])
```
